## 2) Model variance (simple linear regression)
Repeat the simple linear regression using the most important feature found above and using 5 different random seeds (= [11,13,17,19,23]) inside the function train_test_split() and write down MAE (for predictions on the test set) in each case. The standard deviation of the MAEs is a measure of the variance for this model (large std = large model variance). A small Signal-to-Noise ratio (= average MAE / std) is also indicative of a larger model variance.
<Br><Br>

```python
def get_mae(seed):
    """
    This function uses as input a seed number (any integer) and returns the MAE error
    after i) splitting the data into training and test sets, ii) training a model with
    the training set, and iii) performing predictions on the test set.

    How to use this function:
    
    # e.g., split the data using a random seed of 11 (the resulting MAE error is sent to variable 'mae':
    
    >>> mae = get_mae(11)
    
    """
    X = data['Highest degree'].values # features (independent variables)
    y = data['Salary'].values # target (property to be predicted)
    X = X.reshape(-1,1) # we always need a 2D array of features to train a ML model!
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=seed)
    
    # create the model
    lr = LinearRegression() # Create the model
    lr.fit(X_train, y_train) # Train the model with the training set
    pred = lr.predict(X_test) # perform predictions on the test set
    
    # get the metrics
    mae = mean_absolute_error(y_test, pred)
    return mae
```

```python
# Now let's apply the defined function to different seeds
maes = [] # generate an empty list to store the different MAE errors
for seed in [11,13,17,19,23]: # iterate over the seeds used
    mae = get_mae(seed) # call the function defined above for each seed
    maes.append(mae) # store the MAE error as a new element of list 'maes'
mae_mean = np.mean(maes) # calculate the mean of all elements in the list maes (via numpy)
mae_std = np.std(maes) # calculate the standard deviation of all elements in the list maes (via numpy)
print('MAE =', mae_mean, '+/-', mae_std) # finally, print results on the screen
print(f'Signal-to-Noise ratio: {mae_mean/mae_std:.1f}') # the higher, the better
```

Use the expanded dataset with the nine features generated previouosly and train different Ridge models (they use L2 regularization) using the training set and under different regularization parameters: alpha = 0.01, 0.1, 0.5, 1, 5, 10, 100 (in the lecture, this parameter was called $\lambda$). What happens with the coefficients of the fitted linear equation upon increasing alpha?  - Part 2. Which regularization parameter gives the best (smallest) MAE error for predictions on the test set and keeping the same random seed used to split the dataset? #### Part 1

```python
# Define the function that calculates MAE and the linear coefficients

def get_mae_coeff(alpha):
    '''
    alpha (float): regularization parameter
    method = Ridge
    
    How to use it:
        e.g., to do a Ridge regression with alpha of 0.1:
        
        >>> mae, ws = get_mae_coeff(alpha = 0.1)
    
    '''
    # Your code here (the last part of the code is already written below!)
    # Define the nine features and the target




    

    # preprocessing
    xscaler = StandardScaler()
    yscaler = StandardScaler()
    X_train_s = xscaler.fit_transform(X_train)
    X_test_s = xscaler.transform(X_test)
    y_train_s = yscaler.fit_transform(y_train.reshape(-1,1))
    
    # create/train the model
    model = Ridge(alpha = alpha) # create a Ridge regressor
    model.fit(X_train_s, y_train_s) # Train the (Ridge) model with the training set

    # predictions on the test set (+ back preprocessing predictions)
    pred_s = model.predict(X_test_s) # perform predictions on the test set
    pred = yscaler.inverse_transform(pred_s.reshape(-1,1))

    # MAE error and trained linear coefficients of the Ridge model
    mae = mean_absolute_error(y_test, pred)
    ws = model.coef_ # 9 coefficients are given here
    
    return mae, ws
```

```python
# Calculate the coefficients for six different alpha values

alphas = [0.01, 0.1, 0.5, 1, 5, 10, 100] # alphas to be investigated each time a new Ridge model is trained
coeffs = [] # create an empty list for the coefficients (each element is another list with the coefficients of one different regression)
for value in alphas: # iterate over the different alpha values of the list alphas
    mae, ws = get_mae_coeff(alpha = value) # run the function to obtain mae (=error) and ws (= the nine coefficients)
    coeffs.append(list(ws.flatten())) # get coefficients as 1D array, convert the array to a list, append it to the list coeffs
    print(f'MAE: {mae}, Ws: {ws}') # show partial results (of each iteration) on the screen
coefficients = np.array(coeffs).reshape(-1,9) # convert the nested list coeffs to a 2D array with n rows and 9 columns
```

```python
# Plotting the evolution of coefficients

plt.figure(figsize=(10, 6))

for i in range(coefficients.shape[1]):  # Iterate through coefficients (columns)
    plt.plot(alphas, coefficients[:, i], marker='o', label=f'Coefficient {i+1}')

plt.xscale('log') # Logarithmic scale for alpha
plt.xlabel('Alpha (Regularization Strength)')
plt.ylabel('Coefficient Value')
plt.title('Evolution of Ridge Regression Coefficients vs. Alpha')
plt.legend(loc='upper right')
plt.grid(True, which="both", ls="--", c='0.7')
```
